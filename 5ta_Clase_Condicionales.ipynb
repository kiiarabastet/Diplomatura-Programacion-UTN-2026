```python
#CLASIFICADOR DE NOTAS
#El usuario ingresa tres notas (Escolar)
# 1) Calcular el promedio
# 2) Indicar: APROBADO si el promedio está entre 7 y 10
#             DESAPROBADO si el promedio es mayor a 4 y menor a 7
#             APLAZADO si es menor a 4
#             NOTA INVÁLIDA si es menor a 1 o mayor a 10
```

```python
#Gemini

nota1 = int(input("Ingrese nota 1: "))

nota2 = int(input("Ingrese nota 2: "))

nota3 = int(input("Ingrese nota 3: "))

suma = (nota1 + nota2 + nota3)
promedio = suma/3
print("Tiene un promedio de: ", promedio)

if (promedio >= 7 or promedio <= 10):
  print("APROBADO")
elif (promedio > 4 and promedio <7):
  print("DESAPROBADO")
elif (promedio < 4):
  print("APLAZADO")
else:
  print("NOTA INVÁLIDA")
```

```python
#CORREGIDO x profesor

nota1 = int(input("Ingrese nota 1: "))

nota2 = int(input("Ingrese nota 2: "))

nota3 = int(input("Ingrese nota 3: "))

sumaNotas = nota1 + nota2 + nota3
promedioNotas = sumaNotas / 3
print(f"{promedioNotas:2f}")   #Como mostrar con menos decimales

if (promedioNotas >= 7 and promedioNotas <= 10):
  print("Aprobado")
elif (promedioNotas >= 4 and promedioNotas < 7):
  print("Desaprobado")
elif (promedioNotas >= 1 and promedioNotas < 4):
  print("Aplazado")
else:
  print("INVÁLIDO")
```

```python
#Tarea para el miercoles
#El usuario ingresa tres notas (Escolar)
# 1) Calcular el promedio (sumamos los 3 valores ingresados por el usuario y al resultado lo dividimos por tres, ya que calculamos el valor de tres valores)
# 2) Indicar: APROBADO si el promedio está entre 7 y 10 y ninguna nota es menor a 4
#             DESAPROBADO si el promedio es mayor a 4 y menor a 7
#             APLAZADO si el promedio es menor a 4 o si alguna nota es menor a 4, sin importar el promedio
#             NOTA INVÁLIDA si es menor a 1 o mayor a 10

#Ejemplo se cragan las siguientes notas: 10, 10 y 3, el promedio es 7.66 == > APLAZADO porque una de sus notas es menor a 4!!

nota1 = int(input("Ingrese nota 1: "))

nota2 = int(input("Ingrese nota 2: "))

nota3 = int(input("Ingrese nota 3: "))

sumaNotas = nota1 + nota2 + nota3
promedioNotas = sumaNotas / 3
print(f"Promedio: {promedioNotas:.2f}")

if (promedioNotas >= 7 and promedioNotas <= 10) and (nota1 >= 4 and nota1 <= 10) and (nota2 >= 4 and nota2 <= 10) and (nota3 >= 4 and nota3 <= 10):
  print("Aprobado")
elif (promedioNotas > 4 and promedioNotas < 7) and (nota1 >= 4 and nota1 <= 10) and (nota2 >= 4 and nota2 <= 10) and (nota3 >= 4 and nota3 <= 10):
    print("Desaprobado")
elif (promedioNotas < 4 or nota1 < 4 or nota2 < 4 or nota3 < 4) and (nota1 >= 1 and nota2 >= 1 and nota3 >= 1):
    print("Aplazado")
else:
  print("Nota inválida")
```
