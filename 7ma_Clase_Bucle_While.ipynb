```python
#FOR : DESDE o PARA -- Repetir una cantidad PREDETERMINADA (conocida) de veces una o más instrucciones...

for i in range(1,10): #in dentro de... range: Rango         (1,10): Desde el 1 al 9, el ùltimo no lo toma!!!
  print(f"Valor de i en cada vuelta {i}")

print("ENTONCES si yo quiero ir del 1 al 10...")
for i in range(1,11):
  print(i)

print("Voy a probar un for con cantidad variable...")

cant = int(input("Ingrese la cantidad de veces que quiee repetir el bucle: "))

for i in range(1,cant):                            #        (1,cant): desde 1 hasta el anterior a cant
  print(i)

print("Primera forma de resolverlo...")
for i in range(0,cant):                             #       (0,10): desde el 0 hasta el anterior a cant
  print(i)

print("Segunda forma de resolverlo...")             #       (0,10): desde el 0 hasta el anterior a cant
for i in range(cant):
  print(i)

print("Tercera forma de resolverlo...")             #       (1,cant+1): desde 1 hasta cant (si pongo cant +1, el ùltimo no lo incluye, el último no lo incluye, entonces llega hasta cant)
for i in range(1, cant+1):
  print(i)
```

```python
#ALGO QUE CONECTA CON LA CLASE DE PROMEDIO

#VALIDAR DATOS!!! USANDO WHILE !!!

cantValores = int(input("Ingrese cantidad de valores a promediar: "))
while cantValores <= 0:
  cantValores = int(input("Ingrese cantidad de valores a promediar (un número MAYOR a cero): "))


#INICIALIZAR CONTADORES Y SUMADORES !!!
sumaNotas = 0

for NotaActual in range(1, cantValores + 1 ):
  nota = int(input(f"Ingrese la nota nro {NotaActual}: "))

  while nota < 1 or nota > 10:
    print("Error: La nota debe ser >= a 1 ó <= a 10")
    nota =int(input(f"Ingrese la nota nro {NotaActual}: "))

  sumaNotas = sumaNotas + nota

promedio = sumaNotas / cantValores
print(f"Promedio: {promedio:.2f}")

print("Fin del programa")
```

```python
#Programar la tabla de multiplicar...
#Dado un número el programa mostrará la tabla de multiplicar (del 0 al 10) de dicho valot.
#Ejemplo
  #iNGRESO 4 MOSTRARÁ:
# 4 * 0 = 0
# 4 * 1 = 4
# 4 * 2 = 8
#...
#4 * 9 = 36
#4 * 10 =40


númeroTabla = int(input("Ingrese el número del cual quiere ver la tabla: "))

for multiplicador in range(11):                           #Cuenta del 0 al 10
  resultado = númeroTabla * multiplicador                 #Según el número elegido multiplica según coresponda dentro del bucle

  print(númeroTabla, "*", multiplicador, "=", resultado)  #Muestra en pantalla la tabla del 0 al 10 de dicho número escogido
```

```python
#2) DADO UN VALOR POSITIVO MAYOR A 2, MOSTRAR LOS NÚMEROS PARES HASTA DICHO VALOR
# EJEMPLO, SI INGRESO 7
#MUESTRA
#2
#4
#6

#Si ingreso 14
#2
#4
#6
#8
#10
#12

valorIngresado = int(input("Ingrese el número del cual quiere ver los números pares anteriores al mismo: "))
while valorIngresado <= 2:
  valorIngresado = int(input("Ingrese el número del cual quiere ver los números pares anteriores al mismo (¡UN NÚMERO > 2!): "))

for número in range(2, valorIngresado):        #Cuenta desde el 2 hasta el último valor par antes de llegar al valor ingresado
  if número % 2 == 0:                          #Si el número del bucle da como resto cero:
    print(número)                              #Mostrar número en pantalla
```

```python
# 1)Resuelto en clase
tabla = int(input("Ingrese tabla a multiplicar: "))

for factor in range(11): #range (0,11)
  producto = tabla * factor
  print(f"{tabla} * {factor} = {producto}")
```
