```python
#ESTRUCTURAS DE CONTROL DE FLUJO

#CONDICIONAL ==>    if (si...)
#                   else (sino...)

#CONDICIONALES ANIDADOS ==> if ...


#El condicional evalua si una concidción es true or false

# El usuario ingresa un valor, quiero saber si es más grande o más chico que 10
# La pregunta a resolver es:
#   1- ¿Es el valor mayor a 10?
#   2- ¿Es el valor menor a 10?

valor = int(input("Ingrese un valor: "))
print(valor)
# < (menor que)  10 < 25
# > (mayor que)  25 > 10

if (valor > 10 ):
  print("El número ingresado es mayor que 10 ")

else if (valor < 10 ):
  print("EL número ingresado es menor que 10 ")

else: (valor = 10 ):
  print("Es igual a 10 ")

print("FIN")
```

```python
#ELIF

valor = int(input("Ingrese un valor: "))
print(valor)

# < (menor que)  10 < 25
# > (mayor que)  25 > 10

if (valor > 10 ):
  print("El número ingresado es mayor que 10 ")

elif (valor < 10 ): #SINO, osea No es MAYOR a 10, entonces tengo dos opciones: o es menor o es igual... Entonces if pregunto por menor

  print("EL número ingresado es menor que 10 ")

else:
  print("Es igual a 10 ")

print("FIN")
```

```python
valor = int(input("Ingrese un valor: "))

if (valor <= 10 ):
  print("Es menor o igual")

else :
  print ("Es mayor a 10")


# IGUAL == (DOBLE IGUAL! == )

if (valor == 10 ):
  print("Es igual a 10")

#DISTINTO    ! NEGACIÓN con lo cual != está negando la igualdad, o sea, es DISTINTO

if (valor != 10 ):
  print("Es distinto a 10")

```
